```python
import pandas as pd

df = pd.read_csv("spambase.csv")

print("Head(First 5 Rows):")
df.head()
```

```python


X = df.drop("class", axis=1)
y = df["class"]


from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x=df["class"])
plt.title("Class Distribution (0 - Not Spam, 1 - Spam)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

features = [ "word_freq_free", "word_freq_money", "word_freq_business", "capital_run_length_total"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot( x="class",y=feature, data=df, ax=axes[i] )
    axes[i].set_title(f"{feature} vs Class")
    axes[i].set_xlabel("Email Type")
    axes[i].set_xticks([0, 1])
    axes[i].set_xticklabels(["Not Spam", "Spam"])

plt.tight_layout()
plt.savefig("boxPlotsImportantFeatures.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(12,9))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.25,
    random_state=42,
    stratify=y_temp
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

import time

lr = LogisticRegression(max_iter=1000)

start = time.time()
lr.fit(X_train, y_train)
train_time_lr = time.time() - start

y_pred_lr = lr.predict(X_test)

acc_lr = accuracy_score(y_test, y_pred_lr)
prec_lr = precision_score(y_test, y_pred_lr)
rec_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)

print("Accuracy:", acc_lr)
print("Precision:", prec_lr)
print("Recall:", rec_lr)
print("F1 Score:", f1_lr)
print("Training Time:", train_time_lr)
```

```python
from sklearn.model_selection import GridSearchCV

lr_param_grid = {
    "penalty": ["l1", "l2"],
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["liblinear", "saga"]
}

lr_grid = GridSearchCV(
    LogisticRegression(max_iter=2000),
    lr_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

lr_grid.fit(X_train, y_train)

print("Best Parameters:", lr_grid.best_params_)
print("Best CV Accuracy:", lr_grid.best_score_)
```

```python
best_lr = lr_grid.best_estimator_

y_pred_lr_tuned = best_lr.predict(X_test)

print("Tuned Accuracy:", accuracy_score(y_test, y_pred_lr_tuned))
print("Tuned Precision:", precision_score(y_test, y_pred_lr_tuned))
print("Tuned Recall:", recall_score(y_test, y_pred_lr_tuned))
print("Tuned F1:", f1_score(y_test, y_pred_lr_tuned))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Predictions from baseline LR
y_pred_lr = lr.predict(X_test)

# Confusion matrix
cm_lr = confusion_matrix(y_test, y_pred_lr)

# Plot
disp_lr = ConfusionMatrixDisplay(confusion_matrix=cm_lr)
disp_lr.plot()
plt.title("Confusion Matrix - Baseline Logistic Regression")
plt.show()
```

```python
# Predictions from tuned LR
y_pred_lr_tuned = best_lr.predict(X_test)

# Confusion matrix
cm_lr_tuned = confusion_matrix(y_test, y_pred_lr_tuned)

# Plot
disp_lr_tuned = ConfusionMatrixDisplay(confusion_matrix=cm_lr_tuned)
disp_lr_tuned.plot()
plt.title("Confusion Matrix - Tuned Logistic Regression")
plt.show()
```

```python
from sklearn.model_selection import learning_curve
import matplotlib.pyplot as plt
import numpy as np

def plot_learning_curve(model, X, y, title):
    train_sizes, train_scores, val_scores = learning_curve(
        model,
        X,
        y,
        cv=5,
        scoring="accuracy",
        train_sizes=np.linspace(0.1, 1.0, 6),
        n_jobs=-1
    )

    train_mean = train_scores.mean(axis=1)
    val_mean = val_scores.mean(axis=1)

    plt.figure()
    plt.plot(train_sizes, train_mean, label="Training Accuracy")
    plt.plot(train_sizes, val_mean, label="Validation Accuracy")
    plt.xlabel("Training Samples")
    plt.ylabel("Accuracy")
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
plot_learning_curve(lr, X_train, y_train,
                    "Learning Curve - Baseline Logistic Regression")
```

```python
plot_learning_curve(best_lr, X_train, y_train,
                    "Learning Curve - Tuned Logistic Regression")
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score
import time

kernels = ["linear", "poly", "rbf", "sigmoid"]

svm_kernel_results = {}

for k in kernels:
    svm = SVC(kernel=k)

    start = time.time()
    svm.fit(X_train, y_train)
    train_time = time.time() - start

    y_pred = svm.predict(X_test)

    svm_kernel_results[k] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred),
        "Training Time": train_time
    }

svm_kernel_results
```

```python
from sklearn.model_selection import GridSearchCV

svm_param_grid = {
    "kernel": ["linear", "poly", "rbf", "sigmoid"],
    "C": [0.1, 1, 10, 100],
    "gamma": ["scale", "auto"],
    "degree": [2, 3, 4]
}

svm_grid = GridSearchCV(
    SVC(),
    svm_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

svm_grid.fit(X_train, y_train)

print("Best Parameters:", svm_grid.best_params_)
print("Best CV Accuracy:", svm_grid.best_score_)
```

```python
best_svm = svm_grid.best_estimator_

y_pred_svm = best_svm.predict(X_test)

print("Tuned SVM Accuracy:", accuracy_score(y_test, y_pred_svm))
print("Tuned SVM F1:", f1_score(y_test, y_pred_svm))
```

```python
from sklearn.model_selection import learning_curve
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import numpy as np

def plot_learning_curve(model, X, y, title):
    train_sizes, train_scores, val_scores = learning_curve(
        model,
        X,
        y,
        cv=5,
        scoring="accuracy",
        train_sizes=np.linspace(0.1, 1.0, 6),
        n_jobs=-1
    )

    train_mean = train_scores.mean(axis=1)
    val_mean = val_scores.mean(axis=1)

    plt.figure()
    plt.plot(train_sizes, train_mean, label="Training Accuracy")
    plt.plot(train_sizes, val_mean, label="Validation Accuracy")
    plt.xlabel("Training Samples")
    plt.ylabel("Accuracy")
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
kernels = ["linear", "poly", "rbf", "sigmoid"]

for k in kernels:
    svm = SVC(kernel=k)
    plot_learning_curve(
        svm,
        X_train,
        y_train,
        f"Learning Curve - SVM ({k.capitalize()} Kernel)"
    )
```

```python
plot_learning_curve(
    best_svm,
    X_train,
    y_train,
    "Learning Curve - Tuned SVM"
)
```

```python
from sklearn.metrics import roc_curve, auc

kernels = ["linear", "poly", "rbf", "sigmoid"]

for k in kernels:
    svm = SVC(kernel=k)
    svm.fit(X_train, y_train)

    scores = svm.decision_function(X_test)

    fpr, tpr, _ = roc_curve(y_test, scores)
    roc_auc = auc(fpr, tpr)

    plt.figure()
    plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}")
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"ROC Curve - SVM ({k.capitalize()} Kernel)")
    plt.legend()
    plt.show()

```

```python
scores_tuned = best_svm.decision_function(X_test)

fpr_tuned, tpr_tuned, _ = roc_curve(y_test, scores_tuned)
auc_tuned = auc(fpr_tuned, tpr_tuned)

plt.figure()
plt.plot(fpr_tuned, tpr_tuned, label=f"AUC = {auc_tuned:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Tuned SVM")
plt.legend()
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Store models
svm_models = {
    "Linear": SVC(kernel="linear"),
    "Polynomial": SVC(kernel="poly"),
    "RBF": SVC(kernel="rbf"),
    "Sigmoid": SVC(kernel="sigmoid"),
    "Tuned": best_svm
}

plt.figure()

for name, model in svm_models.items():

    # Train if not tuned model
    if name != "Tuned":
        model.fit(X_train, y_train)

    # Get decision scores
    scores = model.decision_function(X_test)

    fpr, tpr, _ = roc_curve(y_test, scores)
    roc_auc = auc(fpr, tpr)

    plt.plot(fpr, tpr, label=f"{name} (AUC = {roc_auc:.3f})")

# Diagonal baseline
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - SVM Kernels and Tuned Model")
plt.legend()
plt.show()
```
